# Fitness Landscape Graph Analyzer

Interactive analysis of fitness landscape graphs from GraphML files.

```python
from graph_analyzer import GraphAnalyzer

graph_path = "../../exps/exp-01-reproduce-results/outputs/run1-reproduce-results/azt_108_0.graphml"
ga = GraphAnalyzer(graph_path)
```

```python
ga.summary(top_k=10)
```

```python
fig = ga.plot()
fig.show()
```

```python
# After hovering over a node in the plot above, paste its name here
ga.inspect("P.LKN...K....")
```

```python
mutants = ga.get_group_mutants("P.LKN...K....")
print(f"{len(mutants)} genotypes in this node")
for m, f in sorted(mutants.items(), key=lambda x: -x[1])[:10]:
    print(f"  {m}  {f:.3f}")
```

```python
# Find which node contains a specific genotype
node = ga.find_genotype(".............")
print(f"Wildtype is in node: {node}")
if node:
    ga.inspect(node)
```

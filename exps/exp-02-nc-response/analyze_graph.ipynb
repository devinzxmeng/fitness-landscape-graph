# Fitness Landscape Graph Analyzer

This notebook demonstrates the `GraphAnalyzer` interface, focusing on the
Hamming-distance layout for simple, interpretable graph visualization.

## Setup

Pick a GraphML file and load the analyzer.

```python
import math

from graph_analyzer import GraphAnalyzer, VisConfig

graph_path = "./outputs/robust-neutral-threshold/neutral-threshold-0-27/azt_108_0.graphml"
ga = GraphAnalyzer(graph_path)
```

## Summary

Quick overview of peaks and connection nodes.

```python
ga.summary(top_k=10)
```

## Hamming-Distance Layout

- X-axis: Hamming distance from wildtype
- Y-axis: Fitness

Two x-axis modes are supported:
- `"node"`: Hamming distance of the supernode ID genotype
- `"average"`: Average Hamming distance across group mutants

The wildtype supernode is always placed at x=0.

```python
config = VisConfig()
config.x_mode = "average"
config.figure_width = 1600
config.figure_height = 1100
config.edge_width_scaling = lambda w: max(0.5, min(3.0, math.log1p(w) * 0.3))

fig = ga.plot(config=config)
fig.show()
```

## Alternate X-Mode (Node-ID Hamming)

```python
config = VisConfig()
config.x_mode = "node"
config.figure_width = 1600
config.figure_height = 1100
config.edge_width_scaling = lambda w: max(0.5, min(3.0, math.log1p(w) * 0.3))

fig = ga.plot(config=config)
fig.show()
```
